# Data Frame should be converted to Dictionary before pushing to MongoDB 

```python
data = df.to_dict(orient="records")
# data
```

```python
import os
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())  # finds the .env file in the project root

DB_NAME = "Vehicle_Insurance"
COLLECTION_NAME = "Claims"
CONNECTION_STRING = os.getenv("MONGODB_CONNECTION_STRING")

if not CONNECTION_STRING:
    raise ValueError("MONGODB_CONNECTION_STRING is not set in .env")
```

```python
client = pymongo.MongoClient(CONNECTION_STRING)
data_base = client[DB_NAME]
collection = data_base[COLLECTION_NAME]
```

```python
# Uploading Data to MongoDB
rec = collection.insert_many(data)
```
